```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# 1. Load Dataset
df = pd.read_csv(r"/content/air - Copy (2).csv")
# 2. Create Date Column
df['Date'] = pd.to_datetime(df['Month'])
# 3. Visualization Settings
sns.set(style="whitegrid")
# 4. Air Passengers Over Time
plt.figure(figsize=(12,6))
plt.plot(df['Date'], df['AirPassengers'], color='blue')
plt.title("Air Passengers Over Time")
plt.xlabel("Date")
plt.ylabel("Air Passengers")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
# 5. Air Passengers Distribution
plt.figure(figsize=(8,5))
sns.histplot(df['AirPassengers'].dropna(), bins=20, kde=True)
plt.title("Distribution of Air Passenger Levels")
plt.xlabel("Air Passengers")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
# 6. Monthly Average Air Passengers
monthly_avg = df.groupby('Month')['AirPassengers'].mean()
plt.figure(figsize=(8,5))
monthly_avg.plot(kind='bar', color='orange')
plt.title("Average Monthly Air Passengers")
plt.xlabel("Month")
plt.ylabel("Average Air Passengers")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# 1. Load Iris Dataset
df = pd.read_csv(r"/content/iris - Copy.csv")
# 2. Visualization Settings
sns.set(style="whitegrid")
palette = sns.color_palette("Set2")  # Colors for species
# 3. Melt the DataFrame for Combined Box Plot
numeric_cols = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
df_melt = df.melt(id_vars='species', value_vars=numeric_cols,
                  var_name='Feature', value_name='Value')
# 4. Combined Box Plot
plt.figure(figsize=(12,6))
sns.boxplot(x='Feature', y='Value', hue='species', data=df_melt, palette=palette)
plt.title("Iris Features Distribution by Species")
plt.xlabel("Feature")
plt.ylabel("Value")
plt.legend(title='Species')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# 1. Load Iris Dataset
df = pd.read_csv(r"/content/iris - Copy.csv")
# 2. Visualization Settings
sns.set(style="whitegrid")
palette = sns.color_palette("Set2", n_colors=df['species'].nunique())  # Colors for species, adjusted to match number of unique species
# 3. Melt the DataFrame for Combined Box Plot for Iris features
numeric_cols = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
df_melt = df.melt(id_vars='species', value_vars=numeric_cols,
                  var_name='Feature', value_name='Value')
# 4. Combined Box Plot
plt.figure(figsize=(12,6))
sns.boxplot(x='Feature', y='Value', hue='species', data=df_melt, palette=palette)
plt.title("Iris Features Distribution by Species")
plt.xlabel("Feature")
plt.ylabel("Value")
plt.legend(title='Species')
plt.tight_layout()
plt.show()
```
